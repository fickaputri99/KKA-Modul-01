```python
class Hero:
    def __init__(self, name, hp, attack_power):   # indentasi 4 spasi
        self.name = name                            # indentasi 8 spasi
        self.hp = hp
        self.attack_power = attack_power

    def info(self):                                 # 4 spasi (satu level dengan __init__)
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

# di luar class, tidak perlu indentasi
hero1 = Hero("Layla", 100, 15)     # baris 10: buat hero1 (HP awal 100)
hero2 = Hero("Zilong", 120, 20)     # baris 11: buat hero2 (HP 120)

hero1.hp = 500

hero1.info()                        # baris ini akan tampil HP: 500
hero2.info()
```

```python
class Hero:
    def __init__(self, name, hp, attack_power):
        self.name = name
        self.hp = hp
        self.attack_power = attack_power

    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

    def serang(self, lawan):
        print(f"{self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    def diserang(self, damage):
        self.hp -= damage
        print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")

hero1 = Hero("Layla", 100, 15)
hero2 = Hero("Zilong", 120, 20)

print("Pertarungan Dimulai")
hero1.serang(hero2)
hero2.serang(hero1)
```
