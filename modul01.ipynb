```python
# Latihan 3
class Hero:
    def __init__(self, name, hp, attack_power):
        self.name = name
        self.hp = hp
        self.attack_power = attack_power

    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

    def serang(self, lawan):
        print(f"{self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    def diserang(self, damage):
        self.hp -= damage
        print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")


class Mage(Hero):
    def __init__(self, name, hp, attack_power, mana):
        super().__init__(name, hp, attack_power)
        self.mana = mana

    def info(self):
        print(f"{self.name} [Mage] | HP: {self.hp} | Mana: {self.mana}")

    def skill_fireball(self, lawan):
        if self.mana >= 20:
            print(f"{self.name} menggunakan Fireball ke {lawan.name}!")
            self.mana -= 20
            lawan.diserang(self.attack_power * 2)
        else:
            print(f"{self.name} gagal skill! Mana tidak cukup.")


print("\n--- Update Class Hero ---")

eudora = Mage("Eudora", 80, 30, 100)
balmond = Hero("Balmond", 200, 10)

eudora.info()
eudora.serang(balmond)
eudora.skill_fireball(balmond)
```

Jika super().__init__(name, hp, attack_power) dihapus, maka atribut name, hp, dan attack_power tidak dibuat pada object Mage. Akibatnya ketika eudora.info() mencoba mengakses self.name, muncul error AttributeError: 'Mage' object has no attribute 'name'. Fungsi super() digunakan untuk memanggil constructor class induk sehingga data dan atribut dari class induk dapat diwariskan dan digunakan oleh class anak.

```python
# Latihan 4
class Hero:
    def __init__(self, nama, hp_awal):
        self.nama = nama
        self.__hp = hp_awal

    def get_hp(self):
        return self.__hp

    def set_hp(self, nilai_baru):
        if nilai_baru < 0:
            self.__hp = 0
        elif nilai_baru > 1000:
            print("Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
            self.__hp = 1000
        else:
            self.__hp = nilai_baru

    def diserang(self, damage):
        sisa_hp = self.get_hp() - damage
        self.set_hp(sisa_hp)

        print(
            f"{self.nama} terkena damage {damage}. "
            f"Sisa HP: {self.get_hp()}"
        )


hero1 = Hero("Layla", 100)

hero1.set_hp(-50)

print(hero1.get_hp())

print(f"Mencoba akses paksa: {hero1._Hero__hp}")

```

1. Percobaan Hacking

Ketika kode print(f"Mencoba akses paksa: {hero1._Hero__hp}") dijalankan, nilai HP akan tetap muncul dan tidak menghasilkan error. Hal ini karena Python menggunakan konsep Name Mangling. Atribut __hp secara internal diubah menjadi _Hero__hp, sehingga masih dapat diakses dengan nama tersebut.

Namun, cara tersebut tidak dianjurkan dalam pemrograman yang baik karena melewati konsep enkapsulasi. Data HP seharusnya diakses melalui Getter get_hp() dan diubah melalui Setter set_hp() agar aturan dan validasi tetap berjalan.

2. Uji Validasi

Jika logika if dan elif pada Setter dihapus sehingga hanya menjadi self.__hp = nilai_baru, kemudian menjalankan hero1.set_hp(-100), maka HP Hero akan menjadi -100.

Hal tersebut terjadi karena tidak ada lagi validasi yang mencegah HP bernilai negatif. Oleh karena itu, Setter sangat penting untuk menjaga integritas data dalam game. Setter dapat memberikan aturan terhadap perubahan data, seperti mencegah HP menjadi negatif atau terlalu besar. Dengan demikian, data HP tetap aman dan sesuai dengan aturan permainan.

```python
# Latihan 5
from abc import ABC, abstractmethod

# 1. Interface / Abstract Class
# Ini adalah KONTRAK.
# Semua turunan wajib memiliki method serang() dan info().
class GameUnit(ABC):

    @abstractmethod
    def serang(self, target):
        pass

    @abstractmethod
    def info(self):
        pass

  # 2. Implementasi pada Class Konkret
class Hero(GameUnit):

    def __init__(self, nama):
        self.nama = nama

    # WAJIB dibuat karena sudah dijanjikan oleh GameUnit
    def serang(self, target):
        print(f"Hero {self.nama} menebas {target}!")

    def info(self):
        print(f"Saya adalah Hero: {self.nama}")



class Monster(GameUnit):

    def __init__(self, jenis):
        self.jenis = jenis

    # Implementasi serang versi Monster
    def serang(self, target):
        print(f"Monster {self.jenis} menggigit {target}!")

    def info(self):
        print(f"Saya adalah Monster: {self.jenis}")


# -- Uji Coba --

# unit = GameUnit()  # ERROR!
# Abstract class tidak bisa langsung dibuat menjadi objek.

h = Hero("Alucard")
m = Monster("Serigala")

h.info()
m.info()
```

1. Melanggar Kontrak

Jika method serang() dihapus dari Hero, akan muncul error karena Hero belum memenuhi aturan dari GameUnit. Akibatnya, objek Hero tidak dapat dibuat sebelum method yang diwajibkan dibuat kembali.

2. Mencetak Cetakan

GameUnit tidak bisa dibuat menjadi objek karena merupakan abstract class. Class ini hanya berfungsi sebagai blueprint atau aturan agar class turunannya seperti Hero dan Monster wajib memiliki method serang() dan info().
